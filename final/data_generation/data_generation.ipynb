```python
# python=3.11
# scipy=1.10.1
# numpy=1.23.5

import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import pickle

from ABM_package_radius import *
```

# part 1 (cell migration only)

#### no "drug" (migration rate rm=5), three cell densities (s0=0.003, 0.03, 0.3), one "movie" per cell density:

```python
np.random.seed(0)

cell_den = [0.003, 0.03, 0.3]
cell_den_labels = ['low', 'med', 'high']

num_t = 40

for i, val in enumerate(cell_den):

    #initial proportions of susceptible, infected, and recovered agents (should sum to less than 1)
    s0 = val
    i0 = 0.0
    r0 = 0.0

    #parameters, only rm relevant here
    ri = 0.1 #infection rate
    rr = ri/10.0 #recovery rate
    rm = 5.0 #migration rate
    radius = 1

    #final time point
    T_end = 1
    t = np.linspace(0,T_end/rr,10)

    #simulate ABM
    S,I,R,t,A_list,_ = SIR_ABM_radius(s0, i0, r0, ri, rr, rm, radius, num_t, T_end)

    metadata = {
        "experiment": "no drug",
        "cell density": cell_den_labels[i],
        "movie frames": 41, 
        "xy dimensions": 40,
        "microns per lattice space": 10, # made this up to give real world units
        "time interval (mins)": 5, # made this up to give real world units
    }

    output = {
        "A_list": A_list,
        "metadata": metadata
    }

    with open(f"migration_no_drug_{cell_den_labels[i]}_density.pkl", "wb") as f:
        pickle.dump(output, f)

```
